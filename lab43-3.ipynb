```python
# 8 = 1000 = 1*2^3 + 0*2^2 + 0*2^1 + 0*2^0
# 1101 = 1*2^3 + 1*2^2 + 0*2^1 + 1*2^0 = 8 + 4 + 0 + 1 = 13

# 13 = 1101
# 13 / 2 = 6 remainder 1
# 6 / 2 = 3 remainder 0
# 3 / 2 = 1 remainder 1

# 17 = 10001
# 17 / 2 = 8 remainder 1
# 8 / 2 = 4 remainder 0
```

```python
N = 13
quotient = N // 2
while quotient > 0:
    remainder = quotient % 2
    print(f"quotient = {quotient}, remainder = {remainder}")
    quotient = quotient // 2
```
